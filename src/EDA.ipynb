```python
# Check image directories map with patient ID

patientID = imageMap['Patient ID']

patientID = patientID.str.replace('\\Data\\', '', regex = False)

patientID = set(patientID)

imageDataDirs = set(d.replace('../data/image_Data/', '') for d in iglob('../data/image_Data/*'))

if patientID == imageDataDirs:
    print('PatientID maps to a directory in image_Data')
else:
    print('PatientId does not match directories in image_Data')
```

```python
prefixes = defaultdict(set)

imageMap['Num Images'] = 0

# Find types and count number of patients
# Also count number of images for each patient
for d in iglob('../data/image_Data/*'):
    p_id = d.replace('../data/image_Data/', '')

    match = re.match(r'([a-zA-Z]+)(\d+)', p_id)

    prefixes[match[1]].add(int(match[2]))

    imageCount = sum(1 for _ in iglob(os.path.join(d, '**/*.jpg'), recursive = True))

    imageMap.loc[imageMap['Patient ID'] == f'\\Data\\{p_id}', 'Num Images'] = imageCount


print('Prefixes and counts')
for prefix in prefixes:
    print(f'{prefix}:', len(prefixes[prefix]))

print()

print('Counts of number of patient images')
print(imageMap['Num Images'].value_counts())
```
